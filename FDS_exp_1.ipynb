```python
# 1.a. Analyze the trend of data
# science job postings over the last decade for the given  data on the number of data science job
# postings each year. Use pandas for data manipulation and matplotlib  for visualization.
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
data={'year':[2016,2017,2018,2019,2020,2021,2022,2023,2024,2025],
'postings':[2500,3560,4892,2159,3567,1564,2790,3249,1576,7335]
      }
df=pd.DataFrame(data)
plt.plot(df['year'],df['postings'],marker='o')
plt.xlabel('year')
plt.ylabel('postings')
plt.xticks(df['year'])
plt.title('Job postings over the decade')
plt.show()
```

```python
data={
    'Roles':['Data scientist','Data analyst','Data engineer','chief data officer','ML engineer','Data scientist','Data analyst','Data engineer','BI developer']
}
df=pd.DataFrame(data)
role_counts=df['Roles'].value_counts()
role_counts.plot(kind='bar')
plt.show()
```

```python
structured=[
    ["Alice","25"],
    ["Bob","30"]
]
semi_structured=[
    {"name":"alice","age":25}
]
unstructured="Alice is 25 years old"
print("Structured:",structured)
print("Semi structured:",semi_structured)
print("Unstructured:",unstructured)
```

```python
from cryptography.fernet import Fernet
key=Fernet.generate_key()
cipher=Fernet(key)
message=b"Confidential data"
encrypted=cipher.encrypt(message)
print("Encrypted:",encrypted)
decrypted=cipher.decrypt(encrypted)
print("Decrypted:",decrypted.decode())
```
